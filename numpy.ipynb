##### Basic Array Creation and Manipulation

1. Create a 1D array:

        Create a 1D array of numbers from 0 to 9.

2. Create a 2D array:

        Create a 2D array of shape (3, 3) with all elements as zeros.
        Create a 2D array of shape (3, 3) with all elements as ones.

```python
arr_2 = np.zeros([3,3])
print(f"array of zeros: \n{arr_2}")
arr_3 = np.ones([3,3])
print(f"\narray of ones: \n{arr_3}")
```

##### Array Operations

5. Element-wise operations:

        Create two arrays of shape (3, 3). Perform element-wise addition, subtraction, multiplication, and division on these arrays.

```python
arr_6 = np.random.randint(1,100,(3,3))
print(arr_6)
arr_7 = np.random.randint(1,100, (3,3))
print("\n", arr_7)
```

```python
arr_6 + arr_7
```

```python
arr_6 - arr_7
```

```python
arr_6 * arr_7
```

```python
arr_6 / arr_7
```

6. Matrix multiplication:

        Create two 2D arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.

```python
arr_8 = np.random.randint(1, 100, (2,3))
arr_9 = np.random.randint(1, 100, (3,2))

print(arr_8)
print("\n", arr_9)
```

```python
# Since the shape of the arrays are different, we can't do element wise multiplication. We can do dot product on both the axis as below.
np.dot(arr_8, arr_9)
```

```python
np.dot(arr_9, arr_8)
```

7. Sum and product:

        Create a 1D array and find the sum and product of its elements.

```python
arr_10 = np.random.randint(1, 100, (5))
arr_10
```

```python
print(f"Sum of the array elements: {sum(arr_10)}")
print(f"Product of the array elements: {arr_10.prod()}")
```

##### Indexing and Slicing

8. Accessing elements:

        Create a 2D array and access the element at row 2, column 1.

```python
arr_11 = np.array(np.random.randint(1,100,(5,4)))
print(f"Array created: \n{arr_11}")
print(f"Element sliced: {arr_11[1,0]}")
```

9. Slicing rows and columns:

        Slice the first two rows and the last two columns of a 2D array.

```python
arr_12 = np.array(np.random.randint(1, 100, (4,5)))
print(f"Array created: \n{arr_12}")
print(f"\nFirst 2 rows: \n{arr_12[:2,:]}")
print(f"\nLast 2 columns: \n{arr_12[:,-2:]}")
```

10. Boolean indexing:

        Create a 1D array of numbers from 1 to 10. Extract the elements that are greater than 5.

```python
arr_13 = np.array(np.random.randint(1, 10, 10))
print(f"Array created: {arr_13}")
print(f"Elements greater than 5: {arr_13[arr_13>5]}")
```

##### Reshaping and Resizing

11. Reshape an array:

        Create a 1D array of 12 elements and reshape it into a 2D array of shape (3, 4).

```python
arr_14 = np.array(np.random.randint(1,100,12))
arr_14 = arr_14.reshape(3,4)
arr_14
```

12. Flatten an array:

        Create a 2D array and flatten it into a 1D array.

```python
arr_15 = np.array(np.random.randint(0,100,(3,4)))
print(f"Original array: \n{arr_15}")
print(f"Array after reshaping: \n{arr_15.reshape(-1)}")
```

##### Statistical Operations

13. Mean, median:

        Create an array and find its mean, median,

```python
arr_16 = np.array(np.random.randint(1,100,10))
print(f"Array: \n{arr_16}")
print(f"Mean of the array: \n{arr_16.mean()}")
print(f"Mode of the array: \n{np.median(arr_16)}")
```

14. Sum:

        Create a 2D array and find the sum

```python
arr_17 = np.array(np.random.randint(1,100,(3,4)))
print(f"Array: \n{arr_17}")
print(f"Sum of the array: \n{arr_17.sum()}")
```

##### Broadcasting

15. Broadcasting addition:

        Create a 2D array and a 1D array. Add the 1D array to each row of the 2D array using broadcasting.

```python
arr_18 = np.array(np.random.randint(1,10,(3,4)))
arr_19 = np.array(np.random.randint(1,10,4))

print(f"2D Array: \n{arr_18}")
print(f"1D Array: \n{arr_19}")

print(f"Sum of arrays: \n{arr_18+arr_19}")
```

16. Broadcasting multiplication:

        Create a 2D array and multiply each element by a scalar using broadcasting.

```python
arr_20 = np.array(np.random.randint(1,10,(3,4)))
print(f"2D array: \n{arr_20}")
print(f"Product of 2: \n{arr_20 * 2}")
```

##### Linear Algebra

17. Transpose of a matrix:

        Create a 2D array and find its transpose.

```python
arr_21 = np.array(np.random.randint(1,20,(4,5)))
print(f"2D array: \n{arr_21}")
print(f"Transpose of the array: \n{arr_21.T}")
```

18. Inverse of a matrix:

        Create a 2D array of shape (2, 2) and find its inverse.

```python
arr_22 = np.array(np.random.randint(0,10,(2,2)))
print(f"Array: \n{arr_22}")
print(f"Inverse of the array: \n{1/arr_22}")
```

19. Determinant of a matrix:

        Create a 2D array of shape (3, 3) and find its determinant.

```python
arr_23 = np.array(np.random.randint(1,10,(3,3)))
print(f"3*3 array: \n{arr_23}")
print(f"Determinant of the array: \n{np.linalg.det(arr_23)}")
```
