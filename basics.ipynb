12. Quadratic Equation:

        Write a Python program that solves the quadratic equation ax2+bx+c=0.
        ax^2 + bx + c = 0
        ax2+bx+c=0. 
        Take random values of a,b, c, x

```python
import numpy as np

a = np.random.randint(-100, 100)
b = np.random.randint(-100, 100)
c = np.random.randint(-100, 100)

print(f"a: {a}, b: {b}, c: {c}")

def qdrtc(a, b, c):
    x1 = -b - ((b**2 - 4*(a*c))**0.5)
    x2 = -b + ((b**2 - 4*(a*c))**0.5)
    return (round(x1,2), round(x2,2))

x = qdrtc(a, b, c)

print(f"Possible values of x: {x}")
```

13. BMI Calculator:

        Write a Python program to calculate the Body Mass Index (BMI). Take the height (in meters) and weight (in kilograms) as input from the user, and use the formula: from the intt 

```python
height = float(input("Enter the height: "))
weight = float(input("Enter the weight: "))

def BMI(height, weight):
    bmi = weight / (height)**2
    return round(bmi, 2)

print(f"Given height: {height}")
print(f"Given weight: {weight}")
print(f"Body Mass Index for the given height and weight: {BMI(height, weight)}")
```

```python
year = int(input("Enter the year to find leap year: "))

if year%4==0:
    if  (year%400==0) or (year%100!=0):
        print(f"The year {year} is a leap year.")
    else:
        print(f"The year {year} is not a leap year.")
else:
    print(f"The year {year} is not a leap year.")
```

19. Greatest Common Divisor (GCD):

        Write a Python program to find the GCD of two numbers. Take the two numbers as input from the user.

```python
num_1 = int(input("Enter the first number to find GCD: "))
num_2 = int(input("Enter the second number to find GCD"))

if (num_1 == 0) & (num_2 == 0):
    print("GCD is undefined since both numbers are 0.")
elif (num_1 == 0) or (num_2 == 0):
    print(f"The GCD of the numbers {num_1} and {num_2} is {max(num_1, num_2)}")
else:
    large_num = max(num_1, num_2)
    small_num = min(num_1, num_2)
    mod = large_num % small_num
    while large_num:
        print(f"large_num: {large_num}")
        print(f"small_num: {small_num}")
        print(f"mod: {mod}")
        large_num -= mod
        if (large_num < small_num) or (large_num <= 0) or (mod <= 0):
            break
        mod = large_num % small_num
        
    print(f"The GCD of the numbers {num_1} and {num_2} is {large_num}")
```
