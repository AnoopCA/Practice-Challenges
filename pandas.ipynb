#### Pivot Pandas

```python
import pandas as pd

data = {
    "City": ["New York", "New York", "New York", "Los Angeles", "Los Angeles",
             "Los Angeles", "Chicago", "Chicago", "Chicago", "Houston",
             "Houston", "Houston", "Miami", "Miami", "Miami",
             "Dallas", "Dallas", "Dallas", "Seattle", "Seattle"],
    
    "Date": pd.to_datetime(["2024-02-01", "2024-02-02", "2024-02-03", "2024-02-04", "2024-02-05",
                            "2024-02-06", "2024-02-07", "2024-02-08", "2024-02-09", "2024-02-10",
                            "2024-02-11", "2024-02-12", "2024-02-13", "2024-02-14", "2024-02-15",
                            "2024-02-16", "2024-02-17", "2024-02-18", "2024-02-19", "2024-02-20"]),
    
    "Product": ["Laptop", "Smartphone", "Tablet", "Laptop", "Smartwatch",
                "Tablet", "Headphones", "Laptop", "Smartphone", "Tablet",
                "Smartwatch", "Laptop", "Smartphone", "Tablet", "Headphones",
                "Smartwatch", "Laptop", "Smartphone", "Tablet", "Smartwatch"],
    
    "Sales": [1520, 980, 1350, 750, 1780, 2100, 890, 1250, 670, 1900,
              1200, 1100, 1300, 780, 1650, 2050, 920, 1400, 720, 1850],
    
    "Profit": [300, 120, 250, 90, 400, 500, 130, 270, 80, 420,
               210, 190, 230, 100, 370, 480, 150, 260, 95, 410]
}

df = pd.DataFrame(data)
print(df)
```

1. How would you create a pivot table to show Sales for each City on each Date using the sample DataFrame?

```python
df.pivot(index="Date", columns="City", values="Sales").fillna(0)
```

2. How can you pivot the DataFrame to show both Sales and Profit for each City on each Date?

```python
df.pivot(index="City", columns="Date", values=["Sales", "Profit"]).fillna(0)
```

3. How would you pivot the DataFrame to have Date and Product as the index, City as columns, and Sales as values?

```python
df.pivot(index=["Date", "Product"], columns="City", values="Sales").fillna(0)
```

4. How can you use pivot_table to calculate the average Sales for each Product in each City?

```python
df.pivot_table(index="Product", columns="City", values="Sales", aggfunc="mean", margins=True, margins_name="Total").fillna(0)
```
