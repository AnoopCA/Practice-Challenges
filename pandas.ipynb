#### Pivot Pandas

```python
import pandas as pd
```

```python
data = {
    "City": ["New York", "New York", "New York", "Los Angeles", "Los Angeles",
             "Los Angeles", "Chicago", "Chicago", "Chicago", "Houston",
             "Houston", "Houston", "Miami", "Miami", "Miami",
             "Dallas", "Dallas", "Dallas", "Seattle", "Seattle"],
    
    "Date": pd.to_datetime(["2024-02-01", "2024-02-02", "2024-02-03", "2024-02-04", "2024-02-05",
                            "2024-02-06", "2024-02-07", "2024-02-08", "2024-02-09", "2024-02-10",
                            "2024-02-11", "2024-02-12", "2024-02-13", "2024-02-14", "2024-02-15",
                            "2024-02-16", "2024-02-17", "2024-02-18", "2024-02-19", "2024-02-20"]),
    
    "Product": ["Laptop", "Smartphone", "Tablet", "Laptop", "Smartwatch",
                "Tablet", "Headphones", "Laptop", "Smartphone", "Tablet",
                "Smartwatch", "Laptop", "Smartphone", "Tablet", "Headphones",
                "Smartwatch", "Laptop", "Smartphone", "Tablet", "Smartwatch"],
    
    "Sales": [1520, 980, 1350, 750, 1780, 2100, 890, 1250, 670, 1900,
              1200, 1100, 1300, 780, 1650, 2050, 920, 1400, 720, 1850],
    
    "Profit": [300, 120, 250, 90, 400, 500, 130, 270, 80, 420,
               210, 190, 230, 100, 370, 480, 150, 260, 95, 410]
}

df = pd.DataFrame(data)
print(df.head(10))
```

1. How would you create a pivot table to show Sales for each City on each Date using the sample DataFrame?

```python
df.pivot(index="Date", columns="City", values="Sales").fillna(0)
```

2. How can you pivot the DataFrame to show both Sales and Profit for each City on each Date?

```python
df.pivot(index="City", columns="Date", values=["Sales", "Profit"]).fillna(0)
```

3. How would you pivot the DataFrame to have Date and Product as the index, City as columns, and Sales as values?

```python
df.pivot(index=["Date", "Product"], columns="City", values="Sales").fillna(0)
```

4. How can you use pivot_table to calculate the average Sales for each Product in each City?

```python
df.pivot_table(index="Product", columns="City", values="Sales", aggfunc="mean", margins=True, margins_name="Total").fillna(0)
```

5. Given the sample DataFrame, how would you pivot it to show the count of Product sold in each City on each Date?

6. How can you fill missing values with 0 in a pivot table created from the sample DataFrame?

```python
# We can use "fillna(0)" to fill missing values with 0.

df.pivot_table(index=["Product", "Date"], columns="City", values="Sales", aggfunc="count").fillna(0)
```

7. If the DataFrame had duplicate entries for Date and City, how can you use pivot_table to handle this by calculating the sum of Sales?

```python
# The parameter "aggfunc" will aggregate the duplicates to the mentioned methond such as sum in this case.
df.pivot_table(index=["Product", "Date"], columns="City", values="Sales", aggfunc="sum").fillna(0)
```

8. How can you perform both mean and sum aggregations on Sales for each Product in each City?

```python
df.pivot_table(index="City", columns="Product", values="Sales",aggfunc=["mean", "sum"]).fillna(0)
```

9. How can you create a pivot table with a MultiIndex on both the rows (Date, Product) and columns (City)?

```python
df.pivot_table(index=["Product", "Date"], columns="City", values="Sales").fillna(0)
```

11. How would you create a pivot table to show monthly Sales for each Product given the Date column includes multiple months?

```python
df_pivot_2 = df.pivot_table(index="Product", columns="Date", values="Sales").fillna(0)
df_pivot_2.columns = [col.strftime("%B") if isinstance(col, pd.Timestamp) else col for col in df_pivot_2.columns]
df_pivot_2.groupby(axis=1, level=0).sum()
```

13. How can you create a pivot table to show the count of Sales transactions for each Product segmented by City?

```python
# The sales field is using the metric "units sold". We can use "sum", if the requirement is to get the number of total products sold.
df.pivot_table(index="City", columns="Product", values="Sales", aggfunc="count").fillna(0)
```

14. How can you prepare the DataFrame for a heatmap visualization by pivoting City, Date, and Sales to have City as rows and Date as columns?

15. How can you use a custom aggregation function (e.g., lambda x: x.max() - x.min()) in a pivot table to calculate the range of Sales for each Product in each
City?

```python
df.pivot_table(index="City", columns="Product", values="Sales", aggfunc=lambda x:x.max()-x.min()).fillna(0)
```
